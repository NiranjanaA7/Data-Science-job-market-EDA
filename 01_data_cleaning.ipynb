## Drop Unnecessary Column

````python
df = df.drop(columns="Unnamed: 0")
````

## Check Duplicate Records

If duplicates exist

````python
df = df.drop_duplicates()
````

## Salary Column
Let's inspect the salary column.

````python
df["Salary Estimate"].sample(10)
````

## Remove Invalid Salaries
Glassdoor uses -1 for unavailable salaries.

````python
df = df[df["Salary Estimate"] != "-1"].copy()
````

## Clean Salary Text

**Before**
```
$53K-$91K (Glassdoor est.)
Employer Provided Salary:$120K-$140K
$40-$60 Per Hour
```

**After**
```
$53K-$91K
Employer Provided Salary:$120K-$140K
$40-$60 Per Hour
```

````python
df["Salary Estimate"] = (
    df["Salary Estimate"]
    .str.split("(")
    .str[0]
    .str.strip()
)
````

Check

````python
df["Salary Estimate"].sample(15)
````

## Per Hour Salary

````python
df["Per_Hour"] = (
    df["Salary Estimate"]
    .str.contains("Per Hour", case=False, na=False)
    .astype(int)
)
````

Check

````python
df["Per_Hour"].value_counts()
````

## Employer Provided Salary

````python
df["Employer_Provided"] = (
    df["Salary Estimate"]
    .str.contains(
        "Employer Provided Salary",
        case=False,
        na=False
    )
    .astype(int)
)
````

Check

````python
df["Employer_Provided"].value_counts()
````

## Remove Salary Descriptors
Now remove unnecessary text.

````python
df["Salary Estimate"] = (
    df["Salary Estimate"]
    .str.replace(
        "Employer Provided Salary:",
        "",
        case=False,
        regex=False
    )
    .str.replace(
        "Per Hour",
        "",
        case=False,
        regex=False
    )
    .str.replace(
        "$",
        "",
        regex=False
    )
    .str.replace(
        "K",
        "",
        regex=False
    )
    .str.strip()
)
````

Check

Expected output
```
53-91
120-140
40-60
```

````python
df["Salary Estimate"].sample(15)
````

## Split Salary

````python
salary = (
    df["Salary Estimate"]
    .str.split("-", expand=True)
)
````

Preview

````python
salary.head()
````

## Minimum Salary

````python
df["Min_Salary"] = (
    salary[0]
    .astype(int)
)
````

## Maximum Salary

````python
df["Max_Salary"] = (
    salary[1]
    .fillna(salary[0])
    .astype(int)
)
````

## Average Salary

````python
df["Avg_Salary"] = (
    df["Min_Salary"] +
    df["Max_Salary"]
) / 2
````

## Validate Salary Columns

````python
df[
    [
        "Salary Estimate",
        "Min_Salary",
        "Max_Salary",
        "Avg_Salary",
        "Per_Hour",
        "Employer_Provided"
    ]
].sample(15)
````
